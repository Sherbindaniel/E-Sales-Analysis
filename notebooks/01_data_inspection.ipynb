# 01. Raw Data Inspection & Profiling
This notebook inspects the raw datasets inside the project directory to evaluate their dimensions, schemas, columns, missing values, duplicates, and timestamp availability.


```python
import os
import zipfile
import pandas as pd
import numpy as np

# Check files in raw folders
print('Web Server Log folder contents:', os.listdir('data/raw/web_server_log'))
print('E-Commerce Log folder contents:', os.listdir('data/raw/ecommerce_log'))
```

## Load raw data into pandas


```python
df_ec_raw = pd.read_csv('data/raw/ecommerce_log/E-commerce Website Logs.csv', low_memory=False)
df_ws_raw = pd.read_csv('data/raw/web_server_log/access_logs.csv')
df_servers = pd.read_csv('data/raw/web_server_log/servers.csv')

print('E-Commerce Raw Shape:', df_ec_raw.shape)
print('Web Server Access Logs Raw Shape:', df_ws_raw.shape)
print('Web Server Metadata Shape:', df_servers.shape)
```

## Inspect Column Schemas and Data Types


```python
print('=== E-Commerce Columns ===')
print(df_ec_raw.info())
print('\n=== Web Server Access Logs Columns ===')
print(df_ws_raw.info())
```

## Inspect Missing Values & Duplicates


```python
print('E-Commerce missing values:\n', df_ec_raw.isnull().sum())
print('E-Commerce duplicates:', df_ec_raw.duplicated().sum())

print('\nWeb Server Access Logs missing values:\n', df_ws_raw.isnull().sum())
print('Web Server Access Logs duplicates:', df_ws_raw.duplicated().sum())
```

## Examine the age anomaly in E-Commerce Logs
As we can see, `age` is classified as `object` (string) and has many missing values. Let's see what unique string values it contains.


```python
print('Sample unique age values:', df_ec_raw['age'].dropna().unique()[:20])
non_numeric = pd.to_numeric(df_ec_raw['age'], errors='coerce').isnull() & df_ec_raw['age'].notnull()
print('Number of non-numeric age rows:', non_numeric.sum())
print('Unique non-numeric age characters:', df_ec_raw[non_numeric]['age'].unique())
```
