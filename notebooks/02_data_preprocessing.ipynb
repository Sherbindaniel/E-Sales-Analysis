# 02. Data Preprocessing & Ingestion Pipeline
This notebook cleans the raw datasets, parses timestamps, handles missing data and string anomalies, and saves the cleaned datasets to `data/processed/`.


```python
import sys
import os
sys.path.insert(0, os.path.abspath('../src'))
import pandas as pd
from data_loader import load_raw_ecommerce_data, load_raw_web_server_data
from preprocessing import clean_ecommerce_data, clean_web_server_data
```

## Load raw data


```python
df_ec_raw = load_raw_ecommerce_data()
df_ws_raw, df_servers = load_raw_web_server_data()
```

## Clean and Preprocess
We parse datetimes, drop duplicates, clean the `--` strings in E-Commerce `age` to `NaN`, and join Web Server access logs with server metadata.


```python
df_ec_clean = clean_ecommerce_data(df_ec_raw)
df_ws_clean = clean_web_server_data(df_ws_raw, df_servers)

print('Cleaned E-Commerce age column summary:')
print(df_ec_clean['age'].describe())
print('\nCleaned Web Server joined columns:')
print(df_ws_clean.head(2))
```

## Save processed datasets


```python
os.makedirs('../data/processed', exist_ok=True)
df_ec_clean.to_csv('../data/processed/clean_ecommerce_logs.csv', index=False)
df_ws_clean.to_csv('../data/processed/clean_web_server_logs.csv', index=False)
print('Saved preprocessed files to data/processed/')
```
